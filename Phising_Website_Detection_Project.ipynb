```python
import pandas as pd

# Phishing dataset ka sample URL load kar rahe hain (Online open dataset)
url = "https://raw.githubusercontent.com/shubham4nand/phishing-dataset/main/phishing.csv"
# Agar yeh link kaam na kare toh hum direct sklearn ya sample data bhi use kar sakte hain.
# Filhaal hum ek chhota aur secure sample dataset banakar shuru karte hain taaki error na aaye.

print("Google Colab setup is working perfectly!")
```

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# 1. Ek dummy dataset create kar rahe hain (Features: URL length, Has '@', Has '-', Target: Safe/Phishing)
data = {
    'url_length': [25, 50, 15, 65, 20, 80, 18, 45],
    'has_at_symbol': [0, 1, 0, 1, 0, 1, 0, 1],
    'has_hyphen': [0, 1, 0, 1, 0, 1, 0, 0],
    'is_phishing': [0, 1, 0, 1, 0, 1, 0, 1] # 0 = Safe, 1 = Phishing
}

df = pd.DataFrame(data)

# 2. Features aur Target ko alag karna
X = df[['url_length', 'has_at_symbol', 'has_hyphen']]
y = df['is_phishing']

# 3. Training aur Testing data mein divide karna
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)

# 4. Random Forest Model ko train karna
model = RandomForestClassifier()
model.fit(X_train, y_train)

# 5. Model ki accuracy check karna
y_pred = model.predict(X_test)
print("Model Accuracy:", accuracy_score(y_test, y_pred) * 100, "%")
print("Machine Learning Model Successfully Trained!")
```

```python
# Naye URL ko test karne ka function
def check_url(new_url):
    # Features extract kar rahe hain (wahi logic jo model training ke waqt use kiya tha)
    length = len(new_url)
    at_symbol = 1 if '@' in new_url else 0
    hyphen = 1 if '-' in new_url else 0

    # Model se prediction lena
    prediction = model.predict([[length, at_symbol, hyphen]])

    if prediction[0] == 1:
        print(f"Warning! '{new_url}' ek **Phishing (Unsafe)** website hai!")
    else:
        print(f"'{new_url}' ek **Safe** website hai.")

# Apne hisab se koi bhi URL yahan daal kar check kar sakte hain:
check_url("https://secure-login-update.com@phishing-site.com")
check_url("https://www.google.com")
```
