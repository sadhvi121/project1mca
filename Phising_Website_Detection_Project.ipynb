```python
# Step 1: Import necessary libraries and prepare the dataset
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

# Dummy dataset representing URL features: [Length, At_Symbol, Hyphen]
# 0 = Safe Website, 1 = Phishing Website
X_train = [
    [15, 0, 0],  # Safe URL example
    [45, 1, 1],  # Phishing URL example
    [12, 0, 0],  # Safe URL example
    [50, 1, 1]   # Phishing URL example
]
y_train = [0, 1, 0, 1]

X_test = [
    [14, 0, 0],
    [48, 1, 1]
]
y_test = [0, 1]
```

```python
# Step 2: Initialize and train the Random Forest model
model = RandomForestClassifier(random_state=42)
model.fit(X_train, y_train)
```

```python
# Step 3: Evaluate model performance
y_pred = model.predict(X_test)
print(f"Model Accuracy: {accuracy_score(y_test, y_pred) * 100}%")
print("Machine Learning Model Successfully Trained!")
```

```python
# Step 4: Define the function to test new URLs and run samples
def check_url(new_url):
    # Feature extraction logic
    length = len(new_url)
    at_symbol = 1 if '@' in new_url else 0
    hyphen = 1 if '-' in new_url else 0

    # Making prediction
    prediction = model.predict([[length, at_symbol, hyphen]])

    if prediction[0] == 1:
        print(f"Warning: '{new_url}' is classified as a Phishing (Unsafe) website!")
    else:
        print(f"'{new_url}' is classified as a Safe website.")

# Testing sample URLs
check_url("https://secure-login-update.com@phishing-site.com")
check_url("https://www.google.com")
```
